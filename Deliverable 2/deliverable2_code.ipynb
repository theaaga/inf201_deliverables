```python
def file_reader() -> tuple[str, str]: 
    """This function reads the files, decodes the files and shows which decoding was used."""

    # Get all files from the nested folders
    files = [p for p in Path("d2_data\\2026").rglob("*") if p.is_file()]

    # For each file that was found, decode and give type of encoding used
    for f in files:
        # Verify the files were found correctly
        print(f)

        # Give the encoding that was used
        best_match = from_path(f).best()
        print("Detected Encoding:", best_match.encoding)

        # Decode the file 

        # Read the raw bytes once, so the file is only opened a single time
        raw = f.read_bytes()

        # Try UTF-8
        # Invalid sequences raise an error.
        try:
            return raw.decode("utf-8-sig"), "utf-8"     # "utf-8-sig" strips BOM if present
        except UnicodeDecodeError:
            pass

        # Else decode it as iso
        return raw.decode("iso-8859-1"), "iso-8859-1"


file_reader()

```

```python
def read_file(path: str | Path) -> tuple[str, str]:
    """Read a file and return (text_content, encoding_used).

    Tries UTF-8 first (strict). If the bytes are not valid UTF-8, falls back
    to ISO-8859-1, which can decode any byte sequence.
    """
    path = Path(path)
    raw = path.read_bytes()  # open the file only once

    try:
        # "utf-8-sig" strips a BOM if present and decodes plain UTF-8 as well
        return raw.decode("utf-8-sig"), "utf-8"
    except UnicodeDecodeError:
        return raw.decode("iso-8859-1"), "iso-8859-1"


def read_all_files(root: str | Path) -> dict[Path, tuple[str, str]]:
    """Recursively read every file under root."""
    root = Path(root)
    results = {}
    for p in sorted(root.rglob("*")):
        if p.is_file():
            results[p] = read_file(p)
    return results


if __name__ == "__main__":
    for path, (text, encoding) in read_all_files(Path("d2_data") / "2026").items():
        print(f"{path}: {encoding} ({len(text)} chars)")
```
